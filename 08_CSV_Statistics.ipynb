# Cargar CSV y calcular estadisticas basicas

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("CSVStats").getOrCreate()

# Cargar CSV en DataFrame
df = spark.read.csv("ruta/al/archivo.csv", header=True, inferSchema=True)

df.show()
```

```python
from pyspark.sql.functions import mean, expr, stddev

# Calcular estadisticas
stats_df = df.select(
    mean("columna").alias("media"),
    expr("percentile_approx(columna, 0.5)").alias("mediana"),
    stddev("columna").alias("desviacion_estandar")
)

stats_df.show()
```

En este tutorial se explica como cargar datos desde un archivo CSV a un DataFrame de Spark y calcular estadísticas básicas como la media, mediana y desviación estándar de una columna numérica.
